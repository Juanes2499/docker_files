# Apache Spark - Notebook de Ejemplo

Este notebook demuestra cómo usar Apache Spark en el entorno Docker.

```python
# Importar librerías necesarias
from pyspark.sql import SparkSession
from pyspark.sql.functions import *
from pyspark.sql.types import *
import pandas as pd
import numpy as np
```

```python
# Crear sesión de Spark
spark = SparkSession.builder \
    .appName("Spark Example") \
    .master("spark://spark-master:7077") \
    .getOrCreate()

print(f"Spark version: {spark.version}")
print(f"Spark context: {spark.sparkContext}")
```

```python
# Crear un DataFrame de ejemplo
data = [
    ("Alice", 25, "Engineer"),
    ("Bob", 30, "Manager"),
    ("Charlie", 35, "Developer"),
    ("Diana", 28, "Analyst"),
    ("Eve", 32, "Designer")
]

schema = StructType([
    StructField("name", StringType(), True),
    StructField("age", IntegerType(), True),
    StructField("role", StringType(), True)
])

df = spark.createDataFrame(data, schema)
df.show()
```

```python
# Operaciones básicas con DataFrames
print("Esquema del DataFrame:")
df.printSchema()

print("\nEstadísticas descriptivas:")
df.describe().show()

print("\nFiltrar por edad mayor a 30:")
df.filter(col("age") > 30).show()
```

```python
# Agregaciones
print("Edad promedio por rol:")
df.groupBy("role").agg(avg("age").alias("avg_age")).show()

print("\nConteo por rol:")
df.groupBy("role").count().show()
```

```python
# Guardar DataFrame
df.write.mode("overwrite").parquet("/opt/spark/data/example_data.parquet")
print("DataFrame guardado en formato Parquet")
```

```python
# Leer DataFrame guardado
df_loaded = spark.read.parquet("/opt/spark/data/example_data.parquet")
print("DataFrame cargado desde Parquet:")
df_loaded.show()
```

```python
# Detener la sesión de Spark
spark.stop()
print("Sesión de Spark finalizada")
```
